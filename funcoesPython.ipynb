def nomeFunção (Aqui eu posso ou não ter argumentos/parametros):
    '''
        Aqui dentro eu coloco a documentação da minha função
    '''
    # A sua função realiza as coisas aqui

    return # Pode ou não ter o return que serve para retornar algo onde chamamos a função     

```python
# Função sem argumento e sem return
def sayHello():
    print("Hello people")

sayHello()    
```

```python
# Função com argumento 
def greeting(name):
    print(f"hello {name}")

nome = input("Digite seu nome")

greeting(nome)
   
```

```python
# Função imprima o maior número
def maior (x,y):
    if x > y:
        print (x)
    else:
        print (y)

maior (10,15)            
```

```python
# Função usando return 
def  addNum(num1, num2):
    return num1+num2

resultado = addNum(10,10)
resultado2 = addNum(10,15)

print(resultado)
print(resultado2)
```

```python
def soma (x,y):
    global total
    total = x+y
    print ("Total soma da função = ", total)

# Programa principal

global total
total = 10
soma(3,5)
print("Total código principal =", total)    
```
