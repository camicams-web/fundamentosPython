```python
# Laço While
count = 0
while count <= 5:
   print(f"O valor do contador é {count}")
   count += 1
print("Aqui o while acabou")   
```

```python
count = 0
while count < 10:
   print(f"O valor do contador é {count}")
   count += 1
   continue
   print ("teste do continue")

print("Aqui o while acabou")   
```

```python
# Laço de repetição FOR
for i in range(-10, 13, 2):
    print(i)
```
