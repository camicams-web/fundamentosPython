```python
minhaString = "hello World!"

# Acessar cararacteres individualmente
minhaString [0] #h
minhaString [1] #E
minhaString [2] #L
minhaString [3] #L
minhaString [4] #O

# Concatenar Strings
outraString = " world!"
novaString = minhaString + outraString
print (novaString)

# Repetir String
repetida = minhaString * 1
print(repetida)

# Fatiar Strings
minhaString [0:5] # Hello
minhaString [:4]  #llo
minhaString [ : :2 ] # Imprimi de 2 em 2 casas Hlo
minhaString [-5] #H
minhaString [-1] #O
minhaString [-3] #L
minhaString [::-1]
print(type(minhaString))
```
